Program 1: Create DataFrame and display data

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("FirstProgram") \
    .master("local[*]") \
    .getOrCreate()

data = [
    (1, "Dinesh", 85),
    (2, "Rahul", 90),
    (3, "Aman", 75)
]

columns = ["id", "name", "marks"]

df = spark.createDataFrame(data, columns)

df.show()
df.printSchema()
```

Program 2: Filter students with marks greater than 80

```python
from pyspark.sql.functions import col

result = df.filter(col("marks") > 80)

result.show()
```

Program 3: Select specific columns

```python
df.select("name", "marks").show()
```
