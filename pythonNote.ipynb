```python
df = spark.createDataFrame(data,columns)
df.show()
df.printSchema()
```

```python
columns=[

    "order_id",

    "customer",

    "product",

    "qty",

    "price",

    "city",

    "order_date"

]




```

# Calculate Revenue

```python
from pyspark.sql.functions import col

df = df.withColumn(
    "revenue",
    col("qty") * col("price")
)

df.select("revenue").show()
```

# TOTAL REVENUE

```python
from pyspark.sql.functions import sum, col

df.select(
    sum(col("revenue")).alias("total_revenue")
).show()
```

# TOTAL THE HIGHEST REVENUE PRODUCT

```python
from pyspark.sql.functions import sum, col, desc

df.groupBy("product") \
    .agg(sum("revenue").alias("total_revenue")) \
    .orderBy(desc("total_revenue")) \
    .show()
```

```python
df.groupBy("product") \
    .agg(sum("revenue").alias("total_revenue")) \
    .orderBy(desc("total_revenue")) \
    .limit(1) \
    .show()
```
